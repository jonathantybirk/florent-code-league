```python
from __future__ import annotations

from enum import Enum
from typing import TYPE_CHECKING, NamedTuple
class EntityType(Enum):
    __slots__ = ()

    BUILDER_BOT = "builder_bot"
    CORE = "core"
    GUNNER = "gunner"
    SENTINEL = "sentinel"
    LAUNCHER = "launcher"
    CONVEYOR = "conveyor"
    SPLITTER = "splitter"
    HARVESTER = "harvester"
    BARRIER = "barrier"
```

```python
for e in EntityType:
    print(f"{e}\n{e.name}\n{e.value}\n")
```
